```python
student = {}
for _ in range(int(input())) :
  name = input()
  score = float(input())
  student[name] = score
scores = list(set(student.values()))
scores.sort()
k = scores[1]
lst = []
for name, score in student.items() :
  if score == k :
    lst.append(name)
lst.sort()
for name in lst :
  print(name)
```
